## Classes: defining your own type

A **class** is a blueprint for objects that belong together because they have the same kind of data and behavior. The objects created from a class are called **instances**. Each instance has its own state.

In the example below:

- `NumberPair` is the class. By convention, class names use `CapWords`.
- `__init__` initializes each new instance.
- `self` refers to the particular instance on which a method is operating.
- `first`, `second`, and `description` are **attributes** (data stored by an instance).
- `describe` and `scale` are **methods** (functions defined inside the class).


```python
class NumberPair:
    """Store and work with a pair of numbers."""

    def __init__(self, first, second, description='A pair of numbers'):
        self.description = description
        self.first = first
        self.second = second

    def describe(self):
        return f'{self.description}: {self.first} and {self.second}'

    def scale(self, factor=2):
        self.first *= factor
        self.second *= factor
```

```python
# Calling the class creates an instance.
pair = NumberPair(1, 2, description='My first pair')

type(pair), isinstance(pair, NumberPair)
```

```python
# Read attributes and call a method with dot notation.
pair.first, pair.second, pair.describe()
```

```python
# A method can change the state of an instance.
pair.scale(factor=3)
pair.describe()
```

```python
# A second instance has independent state.
other_pair = NumberPair(10, 20, description='Another pair')

pair.describe(), other_pair.describe()
```

### Class recap

A function describes an operation to perform. A class defines a kind of object by grouping related state (attributes) and behavior (methods).

The class and its instances are different objects: `NumberPair` is the class, while `pair` and `other_pair` are instances of it. Both instances support the same methods, but each stores its own attribute values.
